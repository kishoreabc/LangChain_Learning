### Streamming

```python
from langchain_google_genai import ChatGoogleGenerativeAI # Import model wrapper
import os
from dotenv import load_dotenv # Import for env management
load_dotenv() # Load API keys from .env
```

```python
os.environ["GEMINI_API_KEY"] = os.getenv("GEMINI_API_KEY") # Set key for langchain-google-genai
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash") # Initialize LLM
```

```python
result = llm.stream("What is agent?") # Request streaming response
for text in result: # Iterate through tokens as they arrive
    print(text.content) # Display stream content in real-time
```

### Batch Process

```python
questions = [
    "What is Streamming process?",
    "What is batch process?",
    "what is the diff betw streamming and batch processing?"
] # List of prompts for parallel/sequential batch processing
```

```python
result = llm.batch(questions) # Process multiple prompts in efficient batch call
```

```python
for batch in result: # Iterate through results list from batch processing
    print(batch.content,end="\n\n\n") # Print each response content
```

#### max concurrent process

```python
result = llm.batch(questions,
config={
    "max_concurrency" : 2   # Limit to 2 parallel calls in standard batch processing
})
```

```python
for batch in result: # Iterate through concurrency-limited results
    print(batch.content,end="\n\n\n") # Display content
```
