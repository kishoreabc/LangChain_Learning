```python
from langchain_google_genai import ChatGoogleGenerativeAI # Import Gemini LLM wrapper
from langchain_core.messages import SystemMessage,HumanMessage,AIMessage,ToolMessage # Import message types
import os
from dotenv import load_dotenv # Import for environment variables
load_dotenv() # Load API keys from .env file
```

```python
os.environ["GEMINI_API_KEY"] = os.getenv("GEMINI_API_KEY") # Set API key for LangChain
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash") # Initialize the Flash model
```

```python
from langchain.tools import tool # Import tool decorator
```

```python
@tool
def get_weather(location:str) -> str:
    """Get the weather at the location""" # Docstring is used as tool description for LLM
    return f"Weather at {location} is 32°C"
```

```python
get_tools = {"get_weather":get_weather} # Map tool names to functions for execution
```

```python
llm_with_tools = llm.bind_tools([get_weather]) # Provide tool schema to the LLM
```

```python
messages = [
    SystemMessage("Your are a helpful assistant"),
    HumanMessage("What is the weather in salem now.is it hot or cold?")
    ] # Define the user query
```

```python
response = llm_with_tools.invoke(messages) # Invoke LLM to get tool call request
```

```python
messages.append(response) # Add the message with tool_calls to history
response
```

```python
for tool in response.tool_calls: # Process each tool call requested by LLM
    tool_response = get_tools[tool["name"]].invoke(tool) # Execute the tool
    messages.append(tool_response) # Add tool result to message history
    print(tool_response.content)
final_response = llm_with_tools.invoke(messages) # Get final natural language response
messages.append(final_response) # Store final response
```
