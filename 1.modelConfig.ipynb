### Model Configuration method 1




```python
from langchain_google_genai import ChatGoogleGenerativeAI # Import Google Gemini model
import os
from dotenv import load_dotenv # Import to load .env variables
load_dotenv() # Load variables from .env file
```

```python
os.environ["GEMINI_API_KEY"] = os.getenv("GEMINI_API_KEY") # Ensure API key is in environment
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash") # Method 1: Direct model initialization
```

```python
result = llm.invoke("What is stremming?") # Call the model with a prompt
print(result.content) # Display the response text
```
